```python
import numpy as np
import matplotlib.pyplot as plt

# Faking a dictionary here, but we would need to know what these vectors correspond to
rgb_dict = {"magenta": [40,120,60],
           "blue": [60,50,90]}


# Drawing the colors in a 3D space

color1 = np.array([40,120,60])
color2 = np.array([60,50,90])


def project_colors(color1, color2):
    # Adding to a graph
    ax = plt.figure().add_subplot(projection='3d')
    
    ax.scatter(xs=color1[0], ys=color1[1], zs=color1[2], zdir='z', label="color1") 
    ax.scatter(xs=color2[0], ys=color2[1], zs=color2[2], zdir='z', label="color2") 
    
    # Specifying the origin
    ax.scatter(0,0,0, zdir='z', label="origin", c="black") 
    
    ax.set_xlim(0,150)
    ax.set_ylim(0,150)
    ax.set_zlim(0,150)
    ax.legend()
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    
    
    # Adding the arrows (from origin to the coordinates of the colors)
    ax.quiver(0,0,0,*color1, arrow_length_ratio=0.1, color="black")
    ax.quiver(0,0,0,*color2, arrow_length_ratio=0.1)
    
    ax.view_init(elev=20, azim=-35, roll=0)
    
    plt.show()

def eucledian_distance(vec_a, vec_b):
    # np.linalg.norm((a-b),ord=2) is the same as the operation below
    return np.sqrt(sum((vec_a - vec_b)**2))

def dot_product(vec_a, vec_b):
    return np.dot(vec_a, vec_b)

def manhattan_distance(vec_a, vec_b):
     # np.linalg.norm((a-b),ord=1) is the same as the operation below
    return sum(np.abs(vec_a - vec_b))

def cosine_distance(vec_a, vec_b):
    '''
    Measures the angular distance between the two vectors. 
    A result of 0 indicates identical direction, while larger values indicate more difference.
    '''
    magnitude = np.linalg.norm(vec_a)*np.linalg.norm(vec_b)
    dot = dot_product(vec_a, vec_b)
    return 1 - (dot / magnitude)

def vector_inputs(vec_a, vec_b):
    print(f"These are the colors:{rgb_dict}")
    print(f"This is the eucledian distance between the two vectors: {eucledian_distance(vec_a, vec_b)}")
    print(f"This is the dot product between the two vectors: {dot_product(vec_a, vec_b)}")
    print(f"This is the cosine distance between the two vectors: {cosine_distance(vec_a, vec_b)}")
    print(f"This is the manhattan distance between the two vectors: {manhattan_distance(vec_a, vec_b)}")

project_colors(color1, color2)
vector_inputs(np.array([40,120,60]), np.array([60,50,90]))

# Coming up with a third color that is closer to color1
color3 = np.array([color1[0], color1[1]+10, color1[2]+5])

project_colors(color1, color3)
    
```
